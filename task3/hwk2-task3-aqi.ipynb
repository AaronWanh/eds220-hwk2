In this task you will use [Air Quality Index (AQI)](https://www.airnow.gov/aqi/aqi-basics/) data from the [US Environmental Protection Agency](https://www.epa.gov) to visualize the impact on the AQI of the 2017 [Thomas Fire](https://en.wikipedia.org/wiki/Thomas_Fire) in Santa Barbara County.

```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt

# Read in data
aqi_17 = pd.read_csv('data/daily_aqi_by_county_2017/daily_aqi_by_county_2017.csv')
aqi_18 = pd.read_csv('data/daily_aqi_by_county_2018/daily_aqi_by_county_2018.csv')

# combine the two datasets together
aqi = pd.concat([aqi_17, aqi_18])

# Simplify column names
aqi.columns = (aqi.columns
                  .str.lower()
                  .str.replace(' ','_')
                )

# filter aqi only in Santa Barbara County and only include the column date to number_of_sites_reporting
aqi_sb = aqi[aqi['county_name'] == 'Santa Barbara']
aqi_sb = aqi_sb.loc[:, ['date', 'aqi', 'category', 'defining_parameter', 'defining_site', 'number_of_sites_reporting']]

# change data type for date to datetime and index the dataframe based on the date
aqi_sb['date'] = pd.to_datetime(aqi_sb['date'])
aqi_sb = aqi_sb.set_index('date').sort_index()

# set the rolling average to 5 days and calculate the mean of aqi
rolling_average = aqi_sb['aqi'].rolling(window = '5D').mean()

# add new column holding the rolling average values
aqi_sb['five_day_average'] = rolling_average

# plot a line graph of daily aqi and 5 day aqi rolling average 

aqi_sb.plot(y= ['aqi', 'five_day_average'], kind = 'line')
plt.title('Air Quality Index in 2017-2018')
plt.xlabel('Year')
plt.ylabel('AQI')
plt.legend(['Daily AQI', '5-Day Rolling Average'])
plt.grid(alpha=0.3)
```

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days.

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values.
